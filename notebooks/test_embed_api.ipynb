In this notebook I want to explore the feasibility of using the free tier of the gemini embedding model to embed the processed chunks, as well as incoming user queries "on the fly" without having to use a locally installed model on the host server.

The next thing I would like to test later is to find an easy way to update a faiss index. I think this should be possible.

```python
# testing gemini embedding api

from google import genai
from google.genai import types

client = genai.Client()

texts = ["What is the shape of a river?",
            "Why study mathematics?",
            "A cat is part horse."]

result = client.models.embed_content(
    model="gemini-embedding-001",
    contents=texts,
    config=types.EmbedContentConfig(output_dimensionality=768)
)

print(type(result.embeddings))
```
